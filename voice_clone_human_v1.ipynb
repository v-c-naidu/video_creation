## How to make it sound human

| Technique | How |
|-----------|-----|
| **Breathing pauses** | Use `...` between phrases |
| **Short sentences** | Each on its own line = natural pause |
| **Slower speed** | `speed = 0.78` feels most natural |
| **More diffusion steps** | `nfe_step = 32` for smooth audio |
| **Punctuation** | Commas and periods control rhythm |
| **Avoid long sentences** | Break anything over 15 words into two |

```python
# ── CELL 2 — Restart kernel ───────────────────────────────────────────
import os
os.kill(os.getpid(), 9)
```

```python
# ── CELL 3 — Load model ───────────────────────────────────────────────
import torch, os, time
from f5_tts.api import F5TTS
from pydub import AudioSegment
from IPython.display import Audio, display

if torch.cuda.is_available():
    print(f'🖥️  GPU: {torch.cuda.get_device_name(0)}')
else:
    print('⚠️  No GPU — go to Runtime → Change runtime type → T4 GPU')

model = F5TTS(device='cuda' if torch.cuda.is_available() else 'cpu')
print('✅ Model ready')
```

```python
# ── CELL 4 — Upload voice sample ──────────────────────────────────────
from google.colab import files

print('📂 Upload your voice sample — 6 to 10 seconds, clean audio, no noise')
uploaded = files.upload()
voice_file = list(uploaded.keys())[0]

dur = len(AudioSegment.from_file(voice_file)) / 1000
print(f'⏱️  Sample duration: {dur:.1f}s')

if dur < 6:
    print('⚠️  Too short — use at least 6 seconds for good cloning')
elif dur > 12:
    print('⚠️  Over 12s — model will clip it. Trim to under 12s.')
else:
    print('✅ Perfect length')

display(Audio(voice_file))
print(f'✅ Loaded: {voice_file}')
```

```python
# ── CELL 5 — Generate Human Sounding Speech ───────────────────────────
#
# TIPS FOR NATURAL SOUND:
#   - Use ... for breathing pauses
#   - Keep sentences short, one per line
#   - Commas and periods control rhythm
#   - Don't write more than 15 words per sentence
# ─────────────────────────────────────────────────────────────────────

# ╔══════════════════════════════════════════════════╗
# ║  YOUR TEXT — edit this every time               ║
# ╚══════════════════════════════════════════════════╝
text = """
**Did you know... some stars are actually cannibals?**

Yep. And it's not science fiction.

There are stars called **white dwarfs** that can steal material from a nearby companion star.

The white dwarf's powerful gravity pulls gas away from its neighbor...

and slowly feeds on it.

Over time, this can cause the white dwarf to become incredibly hot and bright.

And here's the crazy part...

if it steals enough material, it can trigger a massive explosion called a **Type Ia supernova**.

So, the next time you look at the stars...

remember—

space is a lot more violent than it looks.

"""

# ╔══════════════════════════════════════════════════╗
# ║  PASTE EXACTLY WHAT YOUR VOICE SAMPLE SAYS      ║
# ╚══════════════════════════════════════════════════╝
ref_text = "Every star you can see is  part of our galaxy the Milky Way and our galaxy contains hundreds of billions of stars. But here is the incredible part. The Milky Way is the only galaxy among hundreds of billions of galaxies."

# ─────────────────────────────────────────────────
# SETTINGS — tweak these for best results
#
#   speed    : 0.75 = slow & deep | 0.85 = natural | 1.0 = normal
#   nfe_step : 16 = fast | 32 = best quality (recommended)
# ─────────────────────────────────────────────────
speed    = 0.75
nfe_step = 16

# Auto-number files so nothing overwrites
i = 1
while os.path.exists(f'clone_{i:02d}.wav'): i += 1
filename = f'clone_{i:02d}.wav'

print('🎙️  Generating...')
start = time.time()

model.infer(
    ref_file  = voice_file,
    ref_text  = ref_text,
    gen_text  = text.strip(),
    file_wave = filename,
    nfe_step  = nfe_step,
    speed     = speed
)

dur = len(AudioSegment.from_wav(filename)) / 1000
print(f'✅ Done in {time.time()-start:.1f}s')
print(f'⏱️  Duration : {dur:.1f}s ({dur/60:.2f} min)')
print(f'📁 Saved as : {filename}')
display(Audio(filename, autoplay=True))
```

```python
# ── CELL 6 — Download ─────────────────────────────────────────────────
from google.colab import files
print(f'⬇️  Downloading {filename}...')
files.download(filename)
```
