````python
# ============================================================
# CELL 10 — REUSABLE QWEN FUNCTION
# ============================================================

def generate_text(
    prompt,
    max_new_tokens=2000,
    temperature=0.4
):

    messages = [

        {
            "role": "user",
            "content": prompt
        }

    ]


    # --------------------------------------------------------
    # Build Qwen chat prompt
    # --------------------------------------------------------

    text = tokenizer.apply_chat_template(

        messages,

        tokenize=False,

        add_generation_prompt=True,

        enable_thinking=False

    )


    # --------------------------------------------------------
    # Tokenize
    # --------------------------------------------------------

    inputs = tokenizer(

        text,

        return_tensors="pt"

    ).to(model.device)


    # --------------------------------------------------------
    # Generate
    # --------------------------------------------------------

    with torch.no_grad():

        output = model.generate(

            **inputs,

            max_new_tokens=max_new_tokens,

            temperature=temperature,

            top_p=0.9,

            do_sample=True

        )


    # --------------------------------------------------------
    # Remove original prompt
    # --------------------------------------------------------

    generated_tokens = output[

        0,

        inputs["input_ids"].shape[1]:

    ]


    result = tokenizer.decode(

        generated_tokens,

        skip_special_tokens=True

    )


    return result.strip()


print(
    "✓ Qwen generation function ready."
)
````

````python
# ============================================================
# CELL 13 — CLEAN AND PARSE JSON
# ============================================================

import re
import json


# ============================================================
# SAFETY CHECK
# ============================================================

if "raw_script" not in globals():

    raise RuntimeError(
        """
raw_script does not exist.

Please run:

CELL 2  → Configuration
CELL 7  → Load Qwen
CELL 10 → generate_text()
CELL 12 → Generate Script

before running CELL 13.
"""
    )


# ============================================================
# JSON EXTRACTION
# ============================================================

def extract_json(text):

    if not text:

        raise ValueError(
            "Qwen returned an empty response."
        )


    text = text.strip()


    # Remove Markdown JSON fences

    text = re.sub(
        r"```json\s*",
        "",
        text,
        flags=re.IGNORECASE
    )


    text = re.sub(
        r"```\s*",
        "",
        text
    )


    # Find JSON object

    start = text.find("{")

    end = text.rfind("}")


    if start == -1 or end == -1:

        raise ValueError(
            "No JSON object found in Qwen output."
        )


    json_text = text[
        start:end + 1
    ]


    # Parse

    return json.loads(
        json_text
    )


# ============================================================
# PARSE RAW SCRIPT
# ============================================================

try:

    script_data = extract_json(
        raw_script
    )


    print("=" * 70)
    print("✓ JSON PARSED SUCCESSFULLY")
    print("=" * 70)
    print()

    print(
        "Title:",
        script_data.get(
            "title",
            "N/A"
        )
    )

    print(
        "Scenes:",
        len(
            script_data.get(
                "scenes",
                []
            )
        )
    )

    print(
        "Duration:",
        script_data.get(
            "total_duration_seconds",
            "N/A"
        ),
        "seconds"
    )


except Exception as e:

    print("=" * 70)
    print("❌ FAILED TO PARSE SCRIPT")
    print("=" * 70)
    print()

    print(
        "Error:",
        e
    )

    print()

    print("RAW QWEN OUTPUT:")
    print("-" * 70)

    print(raw_script)

    raise
````

````python
# ============================================================
# CELL 15 — VALIDATE SCRIPT
# ============================================================

import math


# ============================================================
# EXPECTED SETTINGS
# ============================================================

EXPECTED_SCENE_DURATION = float(
    SCENE_DURATION
)

EXPECTED_SCENES = int(
    NUMBER_OF_SCENES
)

EXPECTED_TOTAL_DURATION = float(
    TARGET_DURATION_SECONDS
)

TIMING_TOLERANCE = 0.15


# ============================================================
# REQUIRED FIELDS
# ============================================================

REQUIRED_FIELDS = [

    "scene_number",
    "start_time",
    "end_time",
    "narration",
    "visual_description",
    "image_prompt",
    "visual_keyword"

]


# ============================================================
# GET SCENES
# ============================================================

scenes = script_data.get(
    "scenes",
    []
)


validation_errors = []


# ============================================================
# CHECK NUMBER OF SCENES
# ============================================================

if len(scenes) != EXPECTED_SCENES:

    validation_errors.append(

        f"Expected {EXPECTED_SCENES} scenes, "
        f"but Qwen generated {len(scenes)}."

    )


# ============================================================
# CHECK EACH SCENE
# ============================================================

for index, scene in enumerate(
    scenes,
    start=1
):

    # --------------------------------------------------------
    # Required fields
    # --------------------------------------------------------

    for field in REQUIRED_FIELDS:

        if field not in scene:

            validation_errors.append(

                f"Scene {index}: "
                f"Missing '{field}'."

            )


    # --------------------------------------------------------
    # Scene number
    # --------------------------------------------------------

    if scene.get(
        "scene_number"
    ) != index:

        validation_errors.append(

            f"Scene {index}: "
            f"Expected scene number {index}, "
            f"got {scene.get('scene_number')}."

        )


    # --------------------------------------------------------
    # Timing
    # --------------------------------------------------------

    try:

        start = float(
            scene["start_time"]
        )

        end = float(
            scene["end_time"]
        )

        duration = end - start


        # --------------------------------------------
        # Scene duration
        # --------------------------------------------

        if abs(
            duration -
            EXPECTED_SCENE_DURATION
        ) > TIMING_TOLERANCE:

            validation_errors.append(

                f"Scene {index}: "
                f"Duration = {duration:.2f}s, "
                f"expected {EXPECTED_SCENE_DURATION:.2f}s."

            )


    except Exception:

        validation_errors.append(

            f"Scene {index}: "
            f"Invalid start/end time."

        )


    # --------------------------------------------------------
    # Narration
    # --------------------------------------------------------

    if not str(
        scene.get(
            "narration",
            ""
        )
    ).strip():

        validation_errors.append(

            f"Scene {index}: "
            f"Narration is empty."

        )


    # --------------------------------------------------------
    # Image prompt
    # --------------------------------------------------------

    if not str(
        scene.get(
            "image_prompt",
            ""
        )
    ).strip():

        validation_errors.append(

            f"Scene {index}: "
            f"Image prompt is empty."

        )


# ============================================================
# CHECK CONTINUOUS TIMING
# ============================================================

for i in range(
    1,
    len(scenes)
):

    previous_end = float(
        scenes[i - 1]["end_time"]
    )

    current_start = float(
        scenes[i]["start_time"]
    )


    if abs(
        previous_end -
        current_start
    ) > TIMING_TOLERANCE:

        validation_errors.append(

            f"Timing discontinuity between "
            f"Scene {i} and Scene {i + 1}: "
            f"{previous_end}s → {current_start}s."

        )


# ============================================================
# CHECK FIRST SCENE START
# ============================================================

if scenes:

    first_start = float(
        scenes[0]["start_time"]
    )


    if abs(
        first_start
    ) > TIMING_TOLERANCE:

        validation_errors.append(

            f"First scene starts at "
            f"{first_start}s instead of 0.0s."

        )


# ============================================================
# CHECK FINAL SCENE END
# ============================================================

if scenes:

    final_end = float(
        scenes[-1]["end_time"]
    )


    if abs(
        final_end -
        EXPECTED_TOTAL_DURATION
    ) > TIMING_TOLERANCE:

        validation_errors.append(

            f"Final scene ends at "
            f"{final_end}s instead of "
            f"{EXPECTED_TOTAL_DURATION}s."

        )


# ============================================================
# CHECK DECLARED TOTAL DURATION
# ============================================================

declared_duration = float(
    script_data.get(
        "total_duration_seconds",
        EXPECTED_TOTAL_DURATION
    )
)


if abs(
    declared_duration -
    EXPECTED_TOTAL_DURATION
) > TIMING_TOLERANCE:

    validation_errors.append(

        f"Script declares "
        f"{declared_duration}s instead of "
        f"{EXPECTED_TOTAL_DURATION}s."

    )


# ============================================================
# DISPLAY VALIDATION SUMMARY
# ============================================================

print()
print("=" * 80)
print("                    SCRIPT VALIDATION")
print("=" * 80)
print()

print(
    "Expected scenes:",
    EXPECTED_SCENES
)

print(
    "Generated scenes:",
    len(scenes)
)

print(
    "Expected scene duration:",
    EXPECTED_SCENE_DURATION,
    "seconds"
)

print(
    "Expected total duration:",
    EXPECTED_TOTAL_DURATION,
    "seconds"
)

print()


# ============================================================
# RESULT
# ============================================================

if validation_errors:

    print("=" * 80)
    print("❌ VALIDATION FAILED")
    print("=" * 80)
    print()


    for error in validation_errors:

        print(
            "•",
            error
        )


    print()

    print(
        "DO NOT proceed to Cell 16 yet."
    )

    print(
        "Fix/regenerate the script first."
    )


else:

    print("=" * 80)
    print("✅ VALIDATION PASSED")
    print("=" * 80)
    print()

    print(
        "The script is structurally ready "
        "for human review."
    )
````

````python
# ============================================================
# CELL 4 — VALIDATE IMAGE PROMPTS
# ============================================================

REQUIRED_FIELDS = [

    "scene_number",
    "start_time",
    "end_time",
    "duration",
    "visual_keyword",
    "visual_description",
    "image_prompt"

]


validation_errors = []


# ============================================================
# CHECK SCENES
# ============================================================

for expected_number, scene in enumerate(
    image_prompts,
    start=1
):

    # --------------------------------------------------------
    # Required fields
    # --------------------------------------------------------

    for field in REQUIRED_FIELDS:

        if field not in scene:

            validation_errors.append(

                f"Scene {expected_number}: "
                f"Missing field '{field}'."

            )


    # --------------------------------------------------------
    # Scene number
    # --------------------------------------------------------

    if scene.get(
        "scene_number"
    ) != expected_number:

        validation_errors.append(

            f"Expected scene number "
            f"{expected_number}, "
            f"got {scene.get('scene_number')}."

        )


    # --------------------------------------------------------
    # Prompt
    # --------------------------------------------------------

    prompt = str(
        scene.get(
            "image_prompt",
            ""
        )
    ).strip()


    if not prompt:

        validation_errors.append(

            f"Scene {expected_number}: "
            f"image_prompt is empty."

        )


# ============================================================
# RESULT
# ============================================================

print("=" * 80)
print("JSON VALIDATION")
print("=" * 80)
print()


if validation_errors:

    print(
        "❌ VALIDATION FAILED"
    )

    print()

    for error in validation_errors:

        print(
            "•",
            error
        )

    raise ValueError(
        "Fix image_prompts.json before continuing."
    )


else:

    print(
        "✅ image_prompts.json is valid."
    )

    print()

    print(
        "Scenes:",
        len(image_prompts)
    )
````

````python
# ============================================================
# CELL 5 — DISPLAY COMPLETE IMAGE PROMPT PLAN
# ============================================================

print("=" * 100)
print("IMAGE GENERATION PLAN")
print("=" * 100)


for scene in image_prompts:

    print()
    print("=" * 100)

    print(
        f"SCENE {scene['scene_number']:02d}"
    )

    print(
        f"TIME: "
        f"{scene['start_time']}s → "
        f"{scene['end_time']}s"
    )

    print(
        f"DURATION: "
        f"{scene['duration']}s"
    )

    print()

    print("VISUAL KEYWORD:")
    print(
        scene["visual_keyword"]
    )

    print()

    print("VISUAL DESCRIPTION:")
    print(
        scene["visual_description"]
    )

    print()

    print("IMAGE PROMPT:")
    print(
        scene["image_prompt"]
    )


print()
print("=" * 100)

print(
    f"TOTAL SCENES: {len(image_prompts)}"
)

print("=" * 100)
````
